## Definición de la Función Característica

La función característica de la distribución de probabilidad de una variable aleatoria $X$ se define como

$$
\varphi_X(t) = E[e^{itX}]
$$

Sabemos que para cualquier función $g(x)$ de una variable aleatoria $X$, el valor de expectación $E = \int_{-\infty}^{\infty} g(x) f(x) dx$ para el caso continuo, y $E = \sum_{\forall x} g(x) P(x)$ para el caso discreto. Entonces,

### Función Caracterísica Continua

$$
\varphi_X(t) = \int_{-\infty}^{\infty} e^{itx} f_X(x) dx
$$

### Función Caracterísica Discreta

$$
\varphi_X(t) = \sum_{\forall x} e^{itx} P(X = x)
$$

## Propiedades de la Función Característica

1. Siempre continua

2. En probabilidad, permite obtener los momentos de la distribución, por ejemplo, $\varphi_X'(t = 0)$ nos da la media, y $\varphi_X''(t = 0)$ nos da la varianza, sin necesidad de trabajar directamente con la PDF.

## Aplicaciones de la Función Característica

1. Se utilizan para estudiar propiedades de los procesos estocásticos.

2. Ayudan a identificar la distribución subyacente de los datos de series temporales, para elaborar modelos y pronósticos.

3. Permite demostrar el Teorema del Límite Central, al demostrar la convergencia de las funciones características de las sumas con la de la distribución normal.
